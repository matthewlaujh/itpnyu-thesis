```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import umap
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
# Load and preprocess image
def load_image(image_path):
    img = Image.open(image_path)
    img = img.convert('RGB')
    pixels = np.array(img)
    # Reshape to 2D array of pixels
    pixels_2d = pixels.reshape(-1, 3)
    return pixels_2d, img.size
```

```python
# Generate UMAP embedding
def create_umap(pixels, n_neighbors=15, min_dist=0.1, random_state=42):
    # Scale the RGB values
    scaler = StandardScaler()
    scaled_pixels = scaler.fit_transform(pixels)
    
    # Create and fit UMAP
    reducer = umap.UMAP(
        n_neighbors=n_neighbors,
        min_dist=min_dist,
        random_state=random_state
    )
    embedding = reducer.fit_transform(scaled_pixels)
    return embedding
```

```python
# Visualize UMAP results
def plot_umap(embedding, pixels, title="UMAP projection of image colors"):
    plt.figure(figsize=(12, 8))
    plt.scatter(
        embedding[:, 0],
        embedding[:, 1],
        c=pixels/255,
        s=2,
        alpha=0.6
    )
    plt.title(title)
    plt.colorbar()
    plt.show()
```

```python
# Optional: Add clustering to find dominant color groups
def cluster_colors(embedding, n_clusters=5):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    clusters = kmeans.fit_predict(embedding)
    return clusters
```

```python
# Example usage:
# Load your image
image_path = "test.JPG"
pixels, image_size = load_image(image_path)

# Create UMAP embedding
embedding = create_umap(pixels)

# Visualize results
plot_umap(embedding, pixels)

# Optional: Add clustering
clusters = cluster_colors(embedding)
plt.figure(figsize=(12, 8))
plt.scatter(embedding[:, 0], embedding[:, 1], c=clusters, cmap='viridis', s=2)
plt.title("UMAP projection with color clusters")
plt.show()

# Experiment with different parameters
# Try different n_neighbors values
for n in [5, 15, 30]:
    embedding = create_umap(pixels, n_neighbors=n)
    plot_umap(embedding, pixels, f"UMAP with n_neighbors={n}")

# Try different min_dist values
for d in [0.0, 0.1, 0.5]:
    embedding = create_umap(pixels, min_dist=d)
    plot_umap(embedding, pixels, f"UMAP with min_dist={d}")
```
